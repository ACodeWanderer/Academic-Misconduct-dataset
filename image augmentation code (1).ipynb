```python
def add_massive_padding(img, pad_ratio=0.5):
    """Add huge padding so NO rotation/translation ever crops text."""
    pad_w = int(img.width * pad_ratio)
    pad_h = int(img.height * pad_ratio)
    return ImageOps.expand(img, border=(pad_w, pad_h, pad_w, pad_h), fill="white")

def augment_images(input_dir, output_dir, n_aug):
    input_dir = Path(input_dir)
    output_dir = Path(output_dir)
    output_dir.mkdir(exist_ok=True, parents=True)

    img_files = [p for p in input_dir.rglob("*.*")
                 if p.suffix.lower() in [".jpg", ".jpeg", ".png"]]

    print(f"Found {len(img_files)} images...")

    for img_path in img_files:
        try:
            img = Image.open(img_path).convert("RGB")
        except:
            continue

        # 🚨 SUPER PADDING (50%)
        padded = add_massive_padding(img, pad_ratio=0.5)

        stem = img_path.stem
        suffix = img_path.suffix

        for i in range(1, n_aug + 1):
            aug_img = apply_augmentations(padded)

            # 🚨 DO NOT crop back tightly
            # Save the padded + augmented image AS IS
            out_name = f"aug_{stem}_{i}{suffix}"
            aug_img.save(output_dir / out_name)

    print("Done!")
```
